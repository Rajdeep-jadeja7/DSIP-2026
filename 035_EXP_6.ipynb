```python
# FIR Low-Pass Filter Design using Hann Window

# Cutoff Frequency = 3 kHz

import numpy as np

import matplotlib.pyplot as plt

from scipy.signal import firwin, freqz

# Step 1: Define filter parameters

Fs = 10000 # Sampling frequency = 10 kHz

Fc = 3000 # Cutoff frequency = 3 kHz

N = 31 # Number of filter taps

# Step 2: Design the FIR low-pass filter

b = firwin(

numtaps=N,

cutoff=Fc,

window='hann',

pass_zero='lowpass',

fs=Fs

)

# Step 3: Calculate frequency response

f, H = freqz(b, worN=2048, fs=Fs)

# Magnitude response

magnitude_db = 20 * np.log10(np.maximum(np.abs(H), 1e-10))
print("Magnitude response (dB):", magnitude_db)

```

```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import firwin, freqz

# ------------------------------------------------
# FIR Filter Design using Windowing Method
# ------------------------------------------------

N = 31
Fs = 10000
window_type = 'hann'

# Function to design and plot FIR filter
def design_fir_filter(filter_type, cutoff, fs, title):
    b = firwin(
        numtaps=N,
        cutoff=cutoff,
        window=window_type,
        pass_zero=filter_type,
        fs=fs
    )

    f, H = freqz(b, worN=2048, fs=fs)

    magnitude_db = 20 * np.log10(
        np.maximum(np.abs(H), 1e-10)
    )

    plt.figure(figsize=(10, 5))
    plt.plot(f, magnitude_db)
    plt.title(title)
    plt.xlabel('Frequency (Hz)')
    plt.ylabel('Magnitude (dB)')
    plt.grid(True)
    plt.tight_layout()
    plt.show()

    print(f"\n{title}")
    print("Filter coefficients:\n", b)

    return b


# ------------------------------------------------
# 1. FIR Low-Pass Filter
# Cutoff = 3 kHz, Fs = 10 kHz
# ------------------------------------------------
lp = design_fir_filter(
    'lowpass',
    3000,
    10000,
    'FIR Low-Pass Filter (3 kHz)'
)


# ------------------------------------------------
# 2. FIR High-Pass Filter
# Cutoff = 3 kHz, Fs = 10 kHz
# ------------------------------------------------
hp = design_fir_filter(
    'highpass',
    3000,
    10000,
    'FIR High-Pass Filter (3 kHz)'
)


# ------------------------------------------------
# 3. FIR Band-Pass Filter
# Cutoffs = 3 kHz and 6 kHz
# Fs increased to 20 kHz
# ------------------------------------------------
bp = design_fir_filter(
    'bandpass',
    [3000, 6000],
    20000,
    'FIR Band-Pass Filter (3 kHz - 6 kHz)'
)


# ------------------------------------------------
# 4. FIR Band-Stop Filter
# Cutoffs = 3 kHz and 6 kHz
# Fs increased to 20 kHz
# ------------------------------------------------
bs = design_fir_filter(
    'bandstop',
    [3000, 6000],
    20000,
    'FIR Band-Stop Filter (3 kHz - 6 kHz)'
)
```
