```python
#Question 1
import numpy as np
import matplotlib.pyplot as plt

start = -3
stop = 6
step = 1

x = np.arange(start, stop + step, step)

signal = np.zeros(len(x))

for i in range(len(signal)):
    if i % 2 == 0:
        signal[i] = 2
    else:
        signal[i] = 3

plt.stem(x, signal)
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.title("Question 1")
plt.grid(True)
plt.show()
```

```python
#question 2
import numpy as np
import matplotlib.pyplot as plt


start = 0
stop = 5
step = 1

#To Generate x-axis
x = np.arange(start, stop + step, step)
signal = np.array([1, 2, 3, 3, 2, 1])

# Plot
plt.step(x, signal, where='post')
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.title("Question 2")
plt.grid(True)
plt.show()
```

```python
# Question 3
t = np.linspace(-2, 2, 1000)
x = np.zeros(len(t))

for i in range(len(t)):
    if t[i] < -1:
        x[i] = -2
    elif t[i] <= 1:
        x[i] = 2 * t[i]
    else:
        x[i] = 2

plt.plot(t, x)
plt.title("Signal 3")
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python
# Question 4 
def unit_step(length, position):
    signal = np.zeros(length)
    signal[position:] = 1
    return signal

start = -2
stop = 10
step = 1

x = np.arange(start, stop + step, step)

u0 = unit_step(len(x), abs(start))
u3 = unit_step(len(x), abs(start) + 3)
u7 = unit_step(len(x), abs(start) + 7)

signal = u0 - u3 - 5 * u7

plt.stem(x, signal)
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.title("Question 4 ")
plt.grid(True)
plt.show()
```

```python
# Question 5 
def unit_impulse(length, position):
    signal = np.zeros(length)
    signal[position] = 1
    return signal

start = -3
stop = 3
step = 1

x = np.arange(start, stop + step, step)

d0 = unit_impulse(len(x), abs(start))
d1 = unit_impulse(len(x), abs(start) + 1)
dm1 = unit_impulse(len(x), abs(start) - 1)

signal = d0 + 3 * d1 + 5 * dm1

plt.stem(x, signal)
plt.xlabel("Time")
plt.ylabel("Amplitude")
plt.title("Question 5")
plt.grid(True)
plt.show()
```
