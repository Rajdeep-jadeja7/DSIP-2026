```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3)) # Creating a small structuring element


img1 = cv2.imread("image1.png", 0)  # IMAGE 1: Black object with white spots

if img1 is None:
    print("Image 1 not found. Check the file name.")
else:
    # Convert the black object into a white foreground
    _, binary1 = cv2.threshold(
        img1, 127, 255, cv2.THRESH_BINARY_INV
    )

    # Opening removes small unwanted white noise
    opening1 = cv2.morphologyEx(
        binary1, cv2.MORPH_OPEN, kernel
    )

    # Closing fills small holes inside the object
    result1 = cv2.morphologyEx(
        opening1, cv2.MORPH_CLOSE, kernel
    )

    # Convert back to the original black-object appearance
    result1 = cv2.bitwise_not(result1)

    plt.figure(figsize=(10, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(img1, cmap="gray")
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(cv2.bitwise_not(opening1), cmap="gray")
    plt.title("After Opening")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(result1, cmap="gray")
    plt.title("After Opening + Closing")
    plt.axis("off")

    plt.tight_layout()
    plt.show()



img2 = cv2.imread("image2.png", 0)   # IMAGE 2: Old damaged photograph

if img2 is None:
    print("Image 2 not found. Check the file name.")
else:
    #Opening reduces small bright marks and scratches
    opening2 = cv2.morphologyEx(
        img2, cv2.MORPH_OPEN, kernel
    )

    # Closing helps fill narrow dark scratches
    result2 = cv2.morphologyEx(
        opening2, cv2.MORPH_CLOSE, kernel
    )

    plt.figure(figsize=(10, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(img2, cmap="gray")
    plt.title("Original Photograph")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(opening2, cmap="gray")
    plt.title("After Opening")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(result2, cmap="gray")
    plt.title("After Opening + Closing")
    plt.axis("off")

    plt.tight_layout()
    plt.show()



img3 = cv2.imread("image3.png", 0)# IMAGE 3: Fingerprint

if img3 is None:
    print("Image 3 not found. Check the file name.")
else:
    # Convert fingerprint into a binary image
    _, binary3 = cv2.threshold(
        img3, 127, 255, cv2.THRESH_BINARY
    )

    # Opening removes isolated white specks
    opening3 = cv2.morphologyEx(
        binary3, cv2.MORPH_OPEN, kernel
    )

    # Closing fills narrow dark cracks in the ridges
    result3 = cv2.morphologyEx(
        opening3, cv2.MORPH_CLOSE, kernel
    )

    plt.figure(figsize=(10, 4))

    plt.subplot(1, 3, 1)
    plt.imshow(binary3, cmap="gray")
    plt.title("Original Fingerprint")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(opening3, cmap="gray")
    plt.title("After Opening")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(result3, cmap="gray")
    plt.title("After Opening + Closing")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
