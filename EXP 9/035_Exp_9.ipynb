```python
import cv2
import numpy as np

# Load the image
image = cv2.imread("image.png")

if image is None:
    raise FileNotFoundError("Could not find image.png")

# Define the Gaussian kernel
kernel_size = (5, 5)
sigma = 1.5

gaussian_1d = cv2.getGaussianKernel(kernel_size[0], sigma)
gaussian_kernel = np.outer(gaussian_1d, gaussian_1d)

# Apply Gaussian smoothing
smoothed_image = cv2.filter2D(image, -1, gaussian_kernel)

# Define the sharpening kernel
sharpening_kernel = np.array([
    [-1, -1, -1],
    [-1,  9, -1],
    [-1, -1, -1]
], dtype=np.float32)

# Apply sharpening to the original image
sharpened_image = cv2.filter2D(image, -1, sharpening_kernel)

# Display images
cv2.imshow("Original Image", image)
cv2.imshow("Gaussian Smoothed Image", smoothed_image)
cv2.imshow("Sharpened Image", sharpened_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

import cv2
import numpy as np

# Load the image
image = cv2.imread("image.png")

if image is None:
    raise FileNotFoundError("Could not find image.png")

# Define the averaging filter kernel size
kernel_size = (5, 5)

# Create the averaging filter kernel
kernel = np.ones(kernel_size, dtype=np.float32) / (
    kernel_size[0] * kernel_size[1]
)

# Apply averaging filter
smoothed_image = cv2.filter2D(image, -1, kernel)

# Display images
cv2.imshow("Original Image", image)
cv2.imshow("Averaging Filter Smoothed Image", smoothed_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

import cv2

# Load the image
image = cv2.imread("image.png")

if image is None:
    raise FileNotFoundError("Could not find image.png")

# Define the median filter kernel size (must be odd)
kernel_size = 5

# Apply median filter
smoothed_image = cv2.medianBlur(image, kernel_size)

# Display images
cv2.imshow("Original Image", image)
cv2.imshow("Median Filter Smoothed Image", smoothed_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

import cv2
import numpy as np

# Load the image
image = cv2.imread("image.png")

if image is None:
    raise FileNotFoundError("Could not find image.png")

# Apply Gaussian smoothing to reduce noise
blurred_image = cv2.GaussianBlur(image, (5, 5), 0)

# Define the sharpening kernel
laplacian_kernel = np.array([
    [ 0, -1,  0],
    [-1,  5, -1],
    [ 0, -1,  0]
], dtype=np.float32)

# Apply the filter
sharpened_image = cv2.filter2D(
    blurred_image, -1, laplacian_kernel
)

# Display images
cv2.imshow("Original Image", image)
cv2.imshow("Gaussian Blurred Image", blurred_image)
cv2.imshow("Spatially Sharpened Image", sharpened_image)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
