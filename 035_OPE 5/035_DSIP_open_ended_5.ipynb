```python

import cv2
import numpy as np
import matplotlib.pyplot as plt
import os

# List of image names
images = [
    "image1.png",
    "image2.png",
    "image3.png",
    "image4.png"
]

# Create a folder to save output images
os.makedirs("Results", exist_ok=True)

# Process each image
for filename in images:

    # Read image in grayscale
    img = cv2.imread(filename, 0)

    if img is None:
        print("Image not found:", filename)
        continue

    # Apply Median Filters
    median3 = cv2.medianBlur(img, 3)
    median5 = cv2.medianBlur(img, 5)
    median7 = cv2.medianBlur(img, 7)

    # Apply Bilateral Filter
    bilateral = cv2.bilateralFilter(img, 7, 50, 50)

    # Display all results
    plt.figure(figsize=(10, 6))

    plt.subplot(2, 3, 1)
    plt.imshow(img, cmap="gray")
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(2, 3, 2)
    plt.imshow(median3, cmap="gray")
    plt.title("Median 3x3")
    plt.axis("off")

    plt.subplot(2, 3, 3)
    plt.imshow(median5, cmap="gray")
    plt.title("Median 5x5")
    plt.axis("off")

    plt.subplot(2, 3, 4)
    plt.imshow(median7, cmap="gray")
    plt.title("Median 7x7")
    plt.axis("off")

    plt.subplot(2, 3, 5)
    plt.imshow(bilateral, cmap="gray")
    plt.title("Bilateral Filter")
    plt.axis("off")

    plt.tight_layout()

    # Save comparison chart
    output_name = "Results/" + filename.replace(".jpg", "_result.png")
    plt.savefig(output_name)

    plt.show()
    plt.close()


print("All images processed!")
```
